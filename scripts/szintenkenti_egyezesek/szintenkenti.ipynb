```python
import pandas as pd
from pathlib import Path
```

```python
input_file = (
    PROCESSED_DATA_DIR
    / "comparison"
    / "level_matches"
    / "01_jelleg_nelkul.csv"
)

# Adatok beolvasása
df = pd.read_csv(
    input_file,
    dtype=str,
    keep_default_na=False
)

print(f"Sorok száma: {len(df):,}")
df.head()
```

```python
csoportositas = (
    df.groupby(
        [
            "település_azonosító",
            "közterület_neve_22",
            "közterület_jellege_22",
            "közterület_jellege_26"
        ],
        dropna=False
    )
    .size()
    .reset_index(name="címek_száma")
)

# Legnagyobb csoportok elöl
csoportositas = csoportositas.sort_values(
    "címek_száma",
    ascending=False
)

csoportositas.head(20)
#print(len(csoportositas))
```

```python
egyszeri_csoportok = csoportositas[
    csoportositas["címek_száma"] == 1
]


osszes_csoport = len(csoportositas)
egyszeri_db = len(egyszeri_csoportok)

print(f"Összes csoport: {osszes_csoport:,}")
print(f"Pontosan egyszer előforduló: {egyszeri_db:,}")
print(f"Arányuk: {egyszeri_db / osszes_csoport * 100:.2f}%")
```
